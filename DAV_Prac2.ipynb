**Practical 2 | Perform Data inspection, cleaning, handling, missing values, duplicate removal and data transformation using pandas**

**Step 1 : Import Pandas**

```python
# Import the pandas library, which is essential for data manipulation and analysis.
import pandas as pd
```

**step 2:Create the Data**

```python
# Define a dictionary named 'data' that holds sample student information.
# This dictionary will be used to create a DataFrame.
data = {
    'ID' : [101, 102, 103, 104, 104, 105,106, 107],
    "Name" : ['Rahul', 'Priya', 'Amit',None, 'Riya', 'Ajay', 'Naresh', 'Jenil'],
    "Age" : [20,21,None,20,20,22,23,28],
    "City" : ['Surat','Navsari','Surat','Bharuch','Bharuch','Navsari', 'Surat', 'Ahmedabad'],
    "Marks" : [85, None, 78,88,88,81,90,93]
}
```

```python
# Create a pandas DataFrame named 'df' from the 'data' dictionary.
# Then, display the entire DataFrame to see its contents.
df = pd.DataFrame(data)
df
```

**Step 3 : Inspect the Dataset**

```python
# Display first 5 rows
df.head()
```

```python
# Display last 5 rows
df.tail()
```

```python
# Display DataFrame summary
df.info()
```

```python
# Display descriptive statistics
df.describe()
```

```python
# Get number of rows and columns
df.shape
```

```python
# Display data types of each column
df.dtypes
```

```python
# Display column names
df.columns
```

**Step 4 : Check Missing Values**

```python
# Check for missing values
df.isnull()
```

```python
# Count missing values per column
df.isnull().sum()
```

```python
# Count total missing values
df.isnull().sum().sum()
```

**Step 5 : Fill Missing Values**

```python
# Fill missing Marks with mean value
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Display the updated DataFrame to see the changes.
df
```

**Fill Missing Age**

```python
# Fill missing Age with mean value
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

**Fill Missing Name**

```python
# Fill missing Name with "Unknown"
df["Name"] = df["Name"].fillna("Unknown")
df
```

**Step 6 : Detect duplicate rows**

```python
# Check for duplicate rows
df.duplicated()
```

```python
# Count total duplicate rows
df.duplicated().sum()
```

**Step 7 : Remove Duplicate Records**

```python
# Remove duplicate rows
df = df.drop_duplicates()
df
```

**Step 8 : Rename a column**

```python
# Rename column 'Marks' to 'Score'
df.rename(columns={"Marks":"Score"}, inplace=True)

# Display the DataFrame with the updated column name.
df
```

**Step 9 : Change Data Type**

```python
# Convert 'Score' data type to integer
df['Score'] = df['Score'].astype(int)

# Display the data types of all columns to confirm the change.
df.dtypes
```

**Step 10 : Add final score column by adding 5 marks**

```python
# Add 'Final_Score' column by adding 5 bonus marks
df['Final_Score'] = df['Score'] + 5
df
```

**Step 11 : Convert text to uppercase**

```python
# Convert 'Name' text to uppercase
df['Name'] = df['Name'].str.upper()
df
```

**Step 12 : Convert text to lowercase**

```python
# Convert 'City' text to lowercase
df['City'] = df['City'].str.lower()
df
```

**Step 13 : Remove Extra Space**

```python
#Remove extra spaces from 'City'
df['City'] = df['City'].str.strip()
df
```

**Step 14 : Sort the dataset**

```python
#Sort dataset by 'Final_Score' descending
df.sort_values(by="Final_Score", ascending=False)
```

**Step 15 : Save the cleaned dataset**

```python
# Save cleaned dataset to CSV file
df.to_csv("clean_student_data.csv", index=False)
print("Dataset saved successfully")
```
