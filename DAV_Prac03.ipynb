```python
# Import the necessary libraries for data manipulation and math operations
import pandas as pd
import numpy as np
```

**Step 1: Create a Dataframe/Records**

```python
# Initialize a dictionary representing the record set of employees
data = {
    "Emp_ID" : [101, 102, 103, 104, 105, 106, 107, 108],
    "Name" : ["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department" : ["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City" : ["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Build a new pandas DataFrame from the defined dictionary
df = pd.DataFrame(data);
```

```python
# Print the structured DataFrame to the screen
df
```

**Step 2: Filter Data**

2.1 Display employee from IT Department

```python
# Extract employees whose assigned department is 'IT'
df[df["Department"] == "IT"]
```

2.2 Display Employees with salary greater than 20000

```python
# Filter out records where the salary exceeds 20,000
df[df['Salary'] > 20000]
```

2.3 Display Employee from Surat

```python
# Get the subset of employees residing in 'Surat'
df[df['City'] == "Surat"]
```

2.4 Apply Multiple Condition

```python
# Multi-criteria query: Find 'IT' staff with earnings over 20,000
df[(df['Department'] == "IT") & (df['Salary'] > 20000)]
```

**Step 3: Sorting Data**

3.1 Sort Salary in Ascending Order

```python
# Arrange the rows in ascending order based on salary
df.sort_values(by="Salary")
```

3.2 Sort Salary in descending order

```python
# Order the records starting from the highest salary to the lowest
df.sort_values(by="Salary", ascending=False)
```

3.3 Sort Department Wise Salary

```python
# Sort by department name first, and then by salary descending inside each group
df.sort_values(by=["Department", "Salary"], ascending=[True,False])
```

**Step 4: Perform Aggregation**

4.1 Calculate Total Salary

```python
# Sum up the salaries of all employees
df["Salary"].sum()
```

4.2 Calculate the avg salary

```python
# Find the mathematical average of the salary values
df["Salary"].mean()
```

4.3 Find Highest Salary

```python
# Retrieve the highest value from the salary column
df['Salary'].max()
```

4.4 Find Lowest Salary

```python
# Retrieve the absolute lowest salary in the dataset
df["Salary"].min()
```

4.5 Count Employees

```python
# Get the count of records using the Emp_ID column
df['Emp_ID'].count()
```

**Step 5:- Perform groupBy Operation**

5.1 Group Employee Based on department and Claculate the avg salary of each department

```python
# Calculate the mean salary grouped for each distinct department
df.groupby('Department')['Salary'].mean()
```

5.2 Calculate the total salary paid by department

```python
# Compute the cumulative salary payout for each department
df.groupby('Department')['Salary'].sum()
```

5.3 Display the highest salary in each department

```python
# Find the top-tier salary amount within each department
df.groupby('Department')['Salary'].max()
```

5.4 Count Employees Department Wise

```python
# Obtain the headcount of employees in each department
df.groupby('Department')['Salary'].count()
```

5.5 Calculate the average salary of employees in each city

```python
# Analyze average salary payouts relative to each city
df.groupby('City')['Salary'].mean()
```

**Step 6 : Multiple Aggregation Function**

```python
# Generate statistical metrics (count, sum, mean, min, max) for each department
df.groupby('Department')['Salary'].agg(['count', 'sum' ,'mean', 'min', 'max'])
```

**Step 7 : save the result**

```python
# Save the dataset's current state to a CSV file on the server disk
df.to_csv("Employee_Data.csv", index=False)
print("Dataset Saved Successfully")
```
